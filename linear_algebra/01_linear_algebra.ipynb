```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
```

```python
A + B
```

```python
A - B
```

```python
B-A
```

```python
A * 23
```

```python
A * B 
```

### Multiplies matrices using row × column rule.
A @ B
 or
np.matmul(A, B)

```python
A @ B  # matrix multiplication
```

```python
 np.matmul(A ,B)
```

### transpose

```python
A.T
```

### Computes the inverse of a square matrix.


```python
np.linalg.inv(A)
```

```python
A
```

### Computes the determinant.


```python

np.linalg.det(A)
```

### rank of the matrix.

```python

np.linalg.matrix_rank(A)

 
```

```python
### Sum of the main diagonal.
np.trace(A)

 
```

### Matrix dot product.


```python
np.dot(A, B)

 
```

### Solve linear equations ,  Ax = b


 

```python
b = np.array([5, 11])

np.linalg.solve(A, b)
```

### eigenvalues and eigenvectors.



```python
values, vectors = np.linalg.eig(A)

print(values)
print(vectors)
```

### matrix Norm


```python
np.linalg.norm(A)
```

### Multi Dot Product

```python
C = np.array([[9,10],[11,12]])

np.linalg.multi_dot([A, B, C])
```
